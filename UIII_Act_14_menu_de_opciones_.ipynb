#### Anbar Fierro NL = 19
#### problema 19: Realizar un ejemplo de menú, donde podemos escoger las distintas opciones hasta que seleccionamos la opción de “Salir”.



```python
# =========================================================
#                    DATOS DEL ALUMNO
# =========================================================
# NL: 19
# Nombre, apellido: Anbar Fierro
# Escuela: CBTIS128
# Especialidad: INTELIGENCIA ARTIFICIAL
# Gpo: 2H
# Nombre sub modulo: ELABORA PROYECTOS CON PROGRAMACIÓN LÓGICA
# Docente: Ing. Eliseo Nava
# =========================================================

# Inicializamos la variable de la opción con un valor que entre al ciclo
opcion = 0

# Ciclo while que se repetirá hasta que el usuario elija la opción 4 (Salir)
while opcion != 4:
    # Mostramos el menú en pantalla
    print("--- MENÚ DE OPCIONES ---")
    print("1. Saludar")
    print("2. Mostrar datos de la materia")
    print("3. Mensaje motivacional")
    print("4. Salir")

    # Solicitamos la opción al usuario
    opcion = int(input("Selecciona una opción (1-4): "))
    print("") # Línea en blanco para que se vea ordenado

    # Evaluamos la opción seleccionada con estructuras condicionales (if-elif-else)
    if opcion == 1:
        print("¡Hola, Anbar! Mucho éxito en tu práctica de Inteligencia Artificial.")
    elif opcion == 2:
        print("Materia: ELABORA PROYECTOS CON PROGRAMACIÓN LÓGICA")
        print("Docente: Ing. Eliseo Nava")
        print("Grupo: 2H - CBTIS128")
    elif opcion == 3:
        print("El éxito es la suma de pequeños esfuerzos repetidos día tras día.")
    elif opcion == 4:
        print("¡Hasta luego! Programa finalizado con éxito.")
    else:
        print("Opción no válida. Por favor, elige un número del 1 al 4.")

    print("-" * 30) # Separador visual para la siguiente vuelta del ciclo
```
